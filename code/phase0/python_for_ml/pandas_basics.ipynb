```python
# Import pandas for table-like data operations
import pandas as pd

# Import numpy for numerical operations used with pandas
import numpy as np
```

```python
# SECTION 1: Creating a DataFrame
# Create a DataFrame (table) from a Python dictionary
df = pd.DataFrame({
    # House size in square feet
    'size': [600, 800, 1000, 1200, 1500],
    # House price in thousand dollars
    'price': [150, 200, 250, 300, 375],
    # Number of bedrooms
    'bedrooms': [1, 2, 2, 3, 3],
    # Whether a garage exists (0 = no, 1 = yes)
    'garage': [0, 0, 1, 1, 1]
})

# Display the full DataFrame
display(df)

# Print row and column count
print('Shape:', df.shape)

# Print all column names
print('Columns:', df.columns.tolist())

# Print datatype of each column
print('dtypes:\n', df.dtypes)
```

```python
# SECTION 2: Exploring Data
# Show only the first 3 rows to quickly inspect data
display(df.head(3))

# Show descriptive statistics for numeric columns
display(df.describe())

# Count missing values in each column
print('Null counts:\n', df.isnull().sum())

# Print raw numpy array of the price column
print('Prices:', df['price'].values)

# Convert first row to dictionary for readable key-value view
print('Row 0:', df.iloc[0].to_dict())
```

```python
# SECTION 3: Filtering Rows
# Filter rows where bedroom count is greater than 1
display(df[df['bedrooms'] > 1])

# Filter rows with both conditions: size >= 1000 and garage == 1
display(df[(df['size'] >= 1000) & (df['garage'] == 1)])
```

```python
# SECTION 4: Derived Columns & Transformations
# Create price per square foot (convert from thousand dollars to dollars)
df['price_per_sqft'] = (df['price'] / df['size'] * 1000).round(2)

# Create a boolean feature: True if house size is 1000 or more
df['is_large'] = df['size'] >= 1000

# Apply log transform to reduce scale effects in price
df['price_log'] = df['price'].apply(lambda x: round(np.log(x), 3))

# Display selected useful columns
display(df[['size', 'price', 'price_per_sqft', 'is_large', 'price_log']])
```

```python
# SECTION 5: GroupBy & Aggregation
# Group by bedroom count and compute mean price for each group
grouped = df.groupby('bedrooms')['price'].mean()

# Print grouped result
print('Avg price by bedrooms:\n', grouped)

# Build a summary table with multiple aggregate metrics
summary = df.groupby('bedrooms').agg(
    # Average price in each bedroom group
    avg_price=('price', 'mean'),
    # Average size in each bedroom group
    avg_size=('size', 'mean'),
    # Number of rows in each bedroom group
    count=('price', 'count')
)

# Display final summary table
display(summary)
```

```python
# SECTION 6: Extracting NumPy Arrays for Models
# Select feature columns and convert to NumPy array (model input)
X = df[['size', 'bedrooms']].values

# Select target column and convert to NumPy array (model output)
y = df['price'].values

# Print feature matrix with shape information
print(f'Feature matrix X (shape {X.shape}):\n{X}')

# Print target vector with shape information
print(f'Target vector y (shape {y.shape}): {y}')
```
