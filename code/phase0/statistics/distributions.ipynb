# Phase 0 — Statistics: Distributions
Learn what distributions are, why they matter in ML, and how to visualize them with Python.

## 📖 Theory
A distribution tells us how values are spread in data.

In plain English:
- It shows which values are common and which are rare.
- It helps us understand data behavior before modeling.
- Many ML assumptions (for tests, scaling, error analysis) depend on distribution shape.

Common distribution shapes:
- Normal (bell curve): symmetric around mean.
- Uniform: all values are equally likely in a range.
- Skewed: one side has a long tail.
- Exponential: many small values, few large values.

## ⚙️ How It Works
1. Collect values from a feature (for example: heights, income, response time).
2. Count how frequently values appear (histogram).
3. Compare the shape with known patterns (normal, skewed, etc.).
4. Decide preprocessing/model steps based on shape.

Example intuition:
- If data is heavily right-skewed (like income), a log transform can make learning easier.

## 📐 Math
Mean (average):
$$\mu = \frac{1}{n}\sum_{i=1}^{n} x_i$$

Variance (spread):
$$\sigma^2 = \frac{1}{n}\sum_{i=1}^{n}(x_i-\mu)^2$$

Standard deviation:
$$\sigma = \sqrt{\sigma^2}$$

Normal distribution density:
$$f(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

## 🎤 Interview Q&A
1. What is a distribution?
Answer: A distribution describes how values are spread and how often they occur.

2. Why should we check distribution before training a model?
Answer: Distribution reveals skew, outliers, and spread, which affect scaling, transformations, and model behavior.

3. What does standard deviation represent?
Answer: It measures how far values typically lie from the mean.

4. Normal vs skewed data: when does it matter?
Answer: It matters for statistical assumptions, distance-based models, and how we interpret errors.

5. Why use histogram and KDE together?
Answer: Histogram shows counts in bins; KDE gives a smooth estimate of underlying density.

## ⚠️ Trap Qs
1. "If mean and median are equal, data must be normal."
Wrong common answer: True.
Correct: Not always. It can be symmetric without being perfectly normal.

2. "More bins always gives a better histogram."
Wrong common answer: True.
Correct: Too many bins can make noise look like signal.

3. "Standard deviation alone is enough to describe any distribution."
Wrong common answer: True.
Correct: Two distributions can share same std but have very different shapes.

## 🧠 1-Liner
A distribution is the shape of your data, and understanding that shape helps you choose better preprocessing and models.

```python
# Import required libraries for numerical work and visualization
import numpy as np
import matplotlib.pyplot as plt

# Set seed for reproducible random results
np.random.seed(42)

# Generate sample datasets
normal_data = np.random.normal(loc=50, scale=10, size=1000)         # bell-shaped
uniform_data = np.random.uniform(low=20, high=80, size=1000)        # flat-ish
exp_data = np.random.exponential(scale=10, size=1000)               # right-skewed

# Utility function to print summary statistics
def summarize(name, data):
    mean_val = np.mean(data)
    median_val = np.median(data)
    std_val = np.std(data)
    print(f"{name} -> mean={mean_val:.2f}, median={median_val:.2f}, std={std_val:.2f}")

# Print quick numeric summary
summarize('Normal', normal_data)
summarize('Uniform', uniform_data)
summarize('Exponential', exp_data)

# Create subplot figure to compare distributions side-by-side
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Plot normal distribution histogram
axes[0].hist(normal_data, bins=30, color='steelblue', edgecolor='white', alpha=0.85)
axes[0].set_title('Normal Distribution')
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Frequency')
axes[0].grid(True, alpha=0.25)

# Plot uniform distribution histogram
axes[1].hist(uniform_data, bins=30, color='mediumseagreen', edgecolor='white', alpha=0.85)
axes[1].set_title('Uniform Distribution')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('Frequency')
axes[1].grid(True, alpha=0.25)

# Plot exponential distribution histogram
axes[2].hist(exp_data, bins=30, color='coral', edgecolor='white', alpha=0.85)
axes[2].set_title('Exponential (Right-Skewed)')
axes[2].set_xlabel('Value')
axes[2].set_ylabel('Frequency')
axes[2].grid(True, alpha=0.25)

# Fit spacing and render
plt.tight_layout()
plt.show()

# Save plot for revision notes
fig.savefig('distributions_comparison.png', dpi=120)
print('Saved figure: distributions_comparison.png')
```
