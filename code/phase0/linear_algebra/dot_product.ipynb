# Phase 0 — Linear Algebra: Dot Product
Focused notebook for one of the most important operations in ML.

## 📖 Theory
Dot product combines two vectors into one number by multiplying corresponding elements and summing.
In ML, this gives weighted sum: prediction = w · x + b.

## ⚙️ How It Works
1. Pair elements position by position.
2. Multiply each pair.
3. Add all products.

## 📐 Math
For vectors a and b:
a · b = Σ (a_i * b_i)

```python
# Import NumPy
import numpy as np
```

```python
# Define weights and features for a simple model
weights = np.array([0.5, 1.2, -0.3])
features = np.array([10, 2, 4])

# Compute dot product
score = np.dot(weights, features)

print('weights =', weights)
print('features =', features)
print('dot product score =', score)

# PRACTICE: Change feature values and observe score changes
```

```python
# Manual dot product to understand each step
products = weights * features
manual_score = np.sum(products)

print('element-wise products =', products)
print('sum of products =', manual_score)
print('matches np.dot =', manual_score == score)

# PRACTICE: Write your own function dot_manual(a, b)
```

```python
# Add bias term like linear models: y = w.x + b
bias = 1.5
prediction = score + bias

print('bias =', bias)
print('prediction =', prediction)

# PRACTICE: Try a negative bias and compare
```

## 🎤 Interview Q&A
1. What is dot product intuition in ML?
Answer: It measures weighted contribution of each feature to a prediction.

2. Why must vectors have same length for dot product?
Answer: Each position must pair with one matching position for multiplication.

3. Where is dot product used in algorithms?
Answer: Linear regression, logistic regression, neural network layers, and similarity calculations.

4. Dot product vs element-wise multiply?
Answer: Element-wise multiply returns a vector; dot product sums that vector to a single scalar.

5. Why is np.dot preferred over loops?
Answer: It is vectorized and optimized in lower-level compiled code, so it is faster and cleaner.

## ⚠️ Trap Qs
1. Is dot product always positive?
Wrong common answer: Yes, because products add up.
Correct: No, it can be negative, zero, or positive depending on signs and directions.

2. Is np.dot(a, b) same as a * b?
Wrong common answer: Yes, both multiply vectors.
Correct: No, a*b is element-wise, np.dot returns one scalar for 1D vectors.

## 🧠 1-Liner
Dot product turns features and weights into one weighted score, which is the backbone of many ML predictions.
