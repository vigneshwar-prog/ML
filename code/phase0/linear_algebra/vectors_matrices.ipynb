## 📖 Theory
- A vector is an ordered list of numbers.
- A matrix is a table of numbers (rows and columns).
- In ML, each row is usually one data sample and each column is one feature.

## ⚙️ How It Works
1. Store data as vectors/matrices.
2. Apply operations (add, scale, transpose).
3. Feed transformed matrix into ML models.

## 📐 Math
- Vector addition: a + b = [a1+b1, a2+b2, ...]
- Scalar multiplication: c*a = [c*a1, c*a2, ...]
- Matrix shape: m x n (m rows, n columns)

```python
# Import NumPy for vector and matrix operations
import numpy as np
```

```python
# Create two vectors
v1 = np.array([2, 4, 6])
v2 = np.array([1, 3, 5])

# Print vectors
print('v1 =', v1)
print('v2 =', v2)

# Vector addition
print('v1 + v2 =', v1 + v2)

# Vector subtraction
print('v1 - v2 =', v1 - v2)

# Scalar multiplication
print('3 * v1 =', 3 * v1)

# PRACTICE: Create v3 = [10, 20, 30] and print v3 / 10
```

```python
# Create a matrix with 3 rows and 2 columns
X = np.array([
    [1, 10],
    [2, 20],
    [3, 30]
])

# Print matrix
print('Matrix X:\n', X)

# Print matrix shape (rows, columns)
print('Shape of X =', X.shape)

# Access first row
print('First row =', X[0])

# Access second column
print('Second column =', X[:, 1])

# PRACTICE: Print third row and first column
```

```python
# Transpose converts rows to columns and columns to rows
XT = X.T

print('X transpose:\n', XT)
print('Shape of X.T =', XT.shape)

# PRACTICE: Verify that X.T.T gives the original matrix X
```

## 🎤 Interview Q&A
1. What is a vector in ML?
Answer: A vector is an ordered list of numbers used to represent one sample or one parameter set.

2. What is matrix shape and why does it matter?
Answer: Shape tells rows x columns, and operations only work when dimensions are compatible.

3. Why are matrices used for datasets?
Answer: They organize many samples and features in one structure that supports fast vectorized operations.

4. What does transpose do?
Answer: It swaps rows and columns; useful for aligning dimensions in formulas.

5. Difference between a Python list and NumPy vector?
Answer: NumPy vectors are faster and support vectorized math operations directly.

## ⚠️ Trap Qs
1. Can you add vectors of different sizes?
Wrong common answer: Yes, Python handles it.
Correct: No, vector sizes must match for element-wise addition.

2. Is transpose the same as inverse?
Wrong common answer: Yes, both flip matrix behavior.
Correct: No, transpose swaps rows/columns; inverse is a different operation and not all matrices have one.

## 🧠 1-Liner
Vectors and matrices are the core data structures of ML, and shape compatibility controls which operations are valid.
